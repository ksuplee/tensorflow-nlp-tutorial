````python
from transformers import pipeline

# 질의응답 파이프라인 생성
# 한국어 질의응답에 더 적합한 모델로 교체합니다.
qa_pipeline = pipeline(
    "question-answering",
    model="monologg/koelectra-base-v3-finetuned-korquad" # 한국어 질의응답 파인튜닝 모델 예시
)

context = """
오늘은 허깅페이스를 배우는 날입니다. 허깅페이스는 오픈소스 기반의 AI 생태계를 구축하며, 다양한 자연어 처리 모델과 도구를 제공합니다.
파이프라인 기능을 활용하면 요약, 번역, 질의응답 같은 다양한 작업을 쉽게 수행할 수 있습니다.
"""

print("컨텍스트:")
print(context)
print("-" * 30)

while True:
    question = input("질문을 입력하세요 (종료하려면 '종료' 입력): ")
    if question.lower() == '종료':
        break

    # 질의응답 수행
    result = qa_pipeline(question=question, context=context)

    # 결과 출력
    print(f"질문: {question}")
    print(f"답변: {result['answer']}")
    print(f"점수: {result['score']:.4f}")
    print("-" * 30)

print("질의응답 세션을 종료합니다.")
````

````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("psyche/KoT5-summarization")
model = AutoModelForSeq2SeqLM.from_pretrained("psyche/KoT5-summarization")

input_text = "오늘은 허깅페이스를 배우는 날입니다. 이 회사는 오픈소스 기반 AI 생태계로 유명합니다."
print("입력문장: \n"+input_text)
inputs = tokenizer("summarize: " + input_text, return_tensors="pt", max_length=512, truncation=True)

summary_ids = model.generate(inputs["input_ids"], max_length=100, num_beams=4, early_stopping=True)
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("\n요약문장:")
print(summary)
````

````python
#방법 1-2. KoBART 요약 모델(한국어 전용) - Pipeline으로 통합 처리
````

````python
from transformers import pipeline

# pipeline을 사용하여 KoT5 요약 모델 로드
# 모델 이름과 리비전을 명시하는 것이 좋습니다.
summarizer_pipeline = pipeline(
    "summarization",
    model="psyche/KoT5-summarization"
    # KoT5 모델의 특정 리비전을 사용하고 싶다면 revision="..." 추가
)
input_text_for_pipeline = "오늘은 허깅페이스를 배우는 날입니다. 파이프라인을 활용하면 요약, 번역, 질의응답 같은 다양한 작업을 쉽게 수행할 수 있습니다. 이 회사는 오픈소스 기반 AI 생태계로 유명합니다.\n"
print("\n입력문장: \n"+input_text_for_pipeline)

# pipeline을 사용하여 요약 생성
# max_length 대신 max_new_tokens 사용 가능 (모델에 따라 다름)
# KoT5는 max_length가 더 적합할 수 있습니다.
summary_result_pipeline = summarizer_pipeline(
    input_text_for_pipeline,
    max_length=100, # 생성될 요약문의 최대 길이
    min_length=10,  # 생성될 요약문의 최소 길이
    do_sample=False # 샘플링 사용 여부
)

# 결과 출력
print("\n요약문장:")
print(summary_result_pipeline[0]['summary_text'])
````

````python
#방법 2. 다국어 요약 모델(mBART 기반) 사용
````

````python
from transformers import pipeline

# mBART-50 다국어 모델 (요약 태스크)
summarizer = pipeline(
    "summarization",
    model="facebook/mbart-large-50-many-to-many-mmt"
)

# 입력할 텍스트
text = (
    "오늘은 허깅페이스를 배우는 날입니다. 파이프라인을 활용하면 요약, 번역, 질의응답 같은 다양한 작업을 쉽게 수행할 수 있습니다."
)

# >>> 한국어 입력을 넣을 때는 반드시 'ko_KR' 토큰을 언어코드로 지정해야 함
summary = summarizer(
    text,
    max_length=24,
    min_length=10,
    do_sample=False,
    forced_bos_token_id=summarizer.tokenizer.lang_code_to_id["ko_KR"]  # 출력 언어를 한국어로 강제
)

print(summary[0]['summary_text'])
````

# 1. 정규 표현식 실습

```python
import re
```

## 1) .기호 : 한 문자 대체

```python
r = re.compile("a.c")   # "  " 안에 입력된 정규 표현식 패턴을 컴퓨터가 미리 해석해서(컴파일) 객체로 만들어 두는 함수
r.search("kkk")         # 괄호 안에 입력된 문자열(여기서는 "kkk") 전체를 쭉 훑어보면서, 앞서 만든 r의 규칙과 일치하는 부분이 단 하나라도 있는지 검색
   # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abc")
```

2) ?기호 : 제로 또는 한 문자 대체

```python
r = re.compile("ab?c")
r.search("abbc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abc")
```

```python
r.search("ac")
```

## 3) *기호 : 제시 문자가 제로 또는 한개 이상 반복 대체

```python
r = re.compile("ab*c")
r.search("a") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("ac")
```

```python
r.search("abc")
```

```python
r.search("abbbbc")
```

## 4) +기호 : 제시 문자가 한개 이상 반복 대체

```python
r = re.compile("ab+c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abc")
```

```python
r.search("abbbbc")
```

##5) ^기호 : 제시된 문자가 전치되는 문자열

```python
r = re.compile("^ab")
r.search("bbc") # 아무런 결과도 출력되지 않는다.
r.search("zab") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abz")
```

## 6) {숫자} 기호 : 제시된 문자가 숫자 만큼 반복 출현

```python
r = re.compile("ab{2}c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
r.search("abc") # 아무런 결과도 출력되지 않는다.
r.search("abbbbbc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abbc")
```

## 7) {숫자1, 숫자2} 기호 : 제시된 문자가 숫자 범위 만큼 반복 출현

```python
r = re.compile("ab{2,8}c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
r.search("abc") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("abbc")
```

```python
r.search("abbbbbbbbc")
```

```python
r.search("abbbbbbbbbc") # 아무런 결과도 출력되지 않는다.
```

## 8) {숫자,} 기호 : 제시된 문자가 숫자 이상 반복 출현

```python
r = re.compile("a{2,}bc")
r.search("bc") # 아무런 결과도 출력되지 않는다.
r.search("aa") # 아무런 결과도 출력되지 않는다.
r.search("aaaaaaaabc") # 매치
```

```python
r.search("aabc")
```

## 9) [] 기호 : '[ ]' 안의 문자가 하나 이상 매치

```python
r = re.compile("[abc]") # [abc]는 [a-c]와 같다.
r.search("zzz") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("a")
```

```python
r.search("aaaaaaa")
```

```python
r.search("baac")
```

```python
r = re.compile("[a-z]")
r.search("AAA") # 아무런 결과도 출력되지 않는다.
r.search("111") # 아무런 결과도 출력되지 않는다.
```

```python
r.search("aBC")
```

## 10) [^문자] 기호 : '[ ]' 안의 문자가 아닌 문자가 매치

```python
r = re.compile("[^abc]")

# 아래의 세 코드는 아무런 결과도 출력되지 않는다.
r.search("a")
r.search("ab")
r.search("b")
```

```python
r.search("d")
```

```python
r.search("1")
```

# 2. 정규 표현식 모듈 함수 예제


## 1) re.match() 와 re.search()의 차이

- search(): 문자열 전체를 훑어보며 규칙에 맞는 패턴이 숨어있는지 찾습니다.  

- match(): 무조건 문자열의 첫 시작 부분이 규칙과 일치하는지만 검사합니다. 시작이 다르면 뒤에 패턴이 숨어있어도 가차 없이 None(없음)을 반환합니다.  

```python
r = re.compile("ab.")
r.match("kkkabc")  #아무런 결과도 출력되지 않는다.
```

```python
r.search("kkkabc")
```

```python
r.match("abckkk")
```

## 2) re.split()

- 문자열을 특정 패턴(정규 표현식)을 기준으로 쪼개어 리스트(List) 형태로 반환해 주는 함수입니다.  

```python
text = "사과 딸기 수박 메론 바나나"
re.split(" ",text)
```

```python
text="""사과
딸기
수박
메론
바나나"""

re.split("\n",text)
```

```python
text = "사과+딸기+수박+메론+바나나"

re.split("\+",text)
```

## 3) re.findall()

- 문자열 전체를 끝까지 훑어서 일치하는 모든 결과물을 빠짐없이 모아 리스트(List) 형태로 줍니다.  

```python
text = """이름 : 김철수
전화번호 : 010 - 1234 - 1234
나이 : 30
성별 : 남"""

re.findall("\d+",text)
```

```python
re.findall("\d+", "문자열입니다.")
```

## 4) re.sub()

- 문자열 안에서 정규 표현식 규칙과 일치하는 부분을 찾아 다른 글자로 교체(Substitute)해 주는 함수입니다.  

```python
text = "Regular expression : A regular expression, regex or regexp[1] (sometimes called a rational expression)[2][3] is, in theoretical computer science and formal language theory, a sequence of characters that define a search pattern."

preprocessed_text = re.sub('[^a-zA-Z]', ' ', text)
print(preprocessed_text)
```

# 3. 정규 표현식 텍스트 전처리 예제

```python
text = """100 John    PROF
101 James   STUD
102 Mac   STUD"""

re.split('\s+', text)    # 문자열로 분리
```

```python
re.findall('\d+',text)
```

```python
re.findall('[A-Z]',text)
```

```python
re.findall('[A-Z]{4}',text)
```

```python
re.findall('[A-Z][a-z]+',text)
```

# 4. 정규 표현식을 이용한 토큰화

```python
from nltk.tokenize import RegexpTokenizer

text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop"

tokenizer1 = RegexpTokenizer("[\w]+")
tokenizer2 = RegexpTokenizer("[\s]+", gaps=True)

print(tokenizer1.tokenize(text))
print(tokenizer2.tokenize(text))
```

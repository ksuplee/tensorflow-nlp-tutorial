```python
import nltk
nltk.download('wordnet')    # 방대한 영어 어휘 데이터베이스인 워드넷(WordNet)을 다운로드
  # 단어들의 유의어, 반의어 정보뿐만 아니라, 단어의 사전적 기본형을 찾아주는 표제어 추출(Lemmatization) 작업(예: 'doing' -> 'do', 'lives' -> 'life')을 수행할 때 반드시 필요
nltk.download('punkt')      # 텍스트를 문장이나 단어 단위로 쪼개는 토큰화(Tokenization) 작업에 사용되는 모델
nltk.download('punkt_tab')  # 최신 버전의 NLTK에서 punkt 토크나이저가 오류 없이 정상적으로 작동하기 위해 추가로 요구되는 보조 데이터 파일
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ['policy', 'doing', 'organization', 'have', 'going', 'love', 'lives', 'fly', 'dies', 'watched', 'has', 'starting']

print('표제어 추출 전 :',words)
print('표제어 추출 후 :',[lemmatizer.lemmatize(word) for word in words])
```

```python
print(lemmatizer.lemmatize('dies', 'v'))
```

```python
print(lemmatizer.lemmatize('watched', 'v'))
```

```python
print(lemmatizer.lemmatize('has', 'v'))
```

### PorterStemmer (포터 스테머)

- PorterStemmer는 파이썬 NLTK 라이브러리에서 제공하는 가장 대표적인 어간 추출(Stemming) 알고리즘입니다.  

- 단어에서 접미사(예: -ing, -ed, -tion 등)를 기계적으로 잘라내어, 단어의 핵심 뼈대인 어간(Root/Stem)만 남기는 역할을 합니다.  

#### 주요 특징

- 단순한 규칙 기반: 단어의 실제 의미나 문맥, 품사를 깊게 따지지 않습니다. "단어 끝이 특정 알파벳 조합으로 끝나면 잘라낸다"라는 정해진 규칙(알고리즘)에 따라 꼬리를 기계적으로 싹둑 자릅니다.  

- 사전에 없는 단어 생성: 규칙에 따라 과감하게 자르다 보니, 추출된 결과물이 실제 영어 사전에 없는 어색한 형태로 변하는 경우가 많습니다.  

- 현재 화면에 있는 코드 실행 결과 예시:  

    - policy -> polici  
    - organization -> organ  

- 장단점: 처리 속도가 매우 빠르고 텍스트 데이터의 종류를 확 줄여주지만, 정교한 의미 분석에는 한계가 있습니다.  

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

stemmer = PorterStemmer()

sentence = "This was not the map we found in Billy Bones's chest, but an accurate copy, complete in all things--names and heights and soundings--with the single exception of the red crosses and the written notes."
tokenized_sentence = word_tokenize(sentence)

print('어간 추출 전 :', tokenized_sentence)
print('어간 추출 후 :',[stemmer.stem(word) for word in tokenized_sentence])
```

### 어간 추출

```python
words = ['formalize', 'allowance', 'electricical']

print('어간 추출 전 :',words)
print('어간 추출 후 :',[stemmer.stem(word) for word in words])
```

```python
from nltk.stem import PorterStemmer
from nltk.stem import LancasterStemmer

porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()

words = ['policy', 'doing', 'organization', 'have', 'going', 'love', 'lives', 'fly', 'dies', 'watched', 'has', 'starting']
print('어간 추출 전 :', words)
print('포터 스테머의 어간 추출 후:',[porter_stemmer.stem(w) for w in words])
print('랭커스터 스테머의 어간 추출 후:',[lancaster_stemmer.stem(w) for w in words])
```

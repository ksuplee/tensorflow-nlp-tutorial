```python
sent = '김철수는 극중 두 인격의 사나이 이광수 역을 맡았다. 철수는 한국 유일의 태권도 전승자를 가리는 결전의 날을 앞두고 10년간 함께 훈련한 사형인 유연재(김광수 분)를 찾으러 속세로 내려온 인물이다.'
```

```python
from konlpy.tag import Okt
okt = Okt()

sent = '김철수는 극중 두 인격의 사나이 이광수 역을 맡았다. 철수는 한국 유일의 태권도 전승자를 가리는 결전의 날을 앞두고 10년간 함께 훈련한 사형인 유연재(김광수 분)를 찾으러 속세로 내려온 인물이다.'
new_sent = sent.replace(" ", '') # 띄어쓰기가 없는 문장 임의로 만들기

# Pykospacing을 사용하려던 unspaced sentence
print('Original unspaced sentence:', new_sent)

# Okt를 사용하여 토큰화 (띄어쓰기된 결과를 얻을 수 있습니다)
okt_spaced_sent = ' '.join(okt.morphs(new_sent))
print('Spaced sentence with Okt:', okt_spaced_sent)

# 비교를 위해 원래 문장도 출력
print('Original spaced sentence:', sent)
```

# 2. SOYNLP

한국어 자연어 처리를 위한 파이썬 라이브러리입니다. 앞서 보셨던 KoNLPy(Okt 등)와 같은 형태소 분석기들과 비슷한 역할을 하지만, 단어를 찾아내는 '방식'이 완전히 다릅니다.

기존 분석기(KoNLPy)와의 결정적 차이  

- KoNLPy (사전 기반): 이미 만들어져 있는 '국어사전'을 바탕으로 단어를 쪼갭니다. 따라서 사전에 없는 신조어, 유행어, 줄임말(예: '존잼쓰', '킹받네', '갑분싸')을 만나면 제대로 인식하지 못하고 이상하게 쪼개버리는 한계(미등록 단어 문제)가 있습니다.  

- SOYNLP (학습 기반): 사전 없이 데이터(텍스트) 자체에서 규칙을 찾아내어(비지도 학습) 단어를 유추합니다. 주어진 텍스트 안에서 특정 글자들이 얼마나 자주 같이 붙어 다니는지를 계산하여 "아, 이 글자 뭉치가 하나의 단어구나!"라고 기계가 스스로 판단하게 합니다.

```python
import urllib.request
from soynlp import DoublespaceLineCorpus
from soynlp.word import WordExtractor
```

```python
urllib.request.urlretrieve("https://raw.githubusercontent.com/lovit/soynlp/master/tutorials/2016-10-20.txt", filename="2016-10-20.txt")
```

```python
# 훈련 데이터를 다수의 문서로 분리
corpus = DoublespaceLineCorpus("2016-10-20.txt")
len(corpus)
```

```python
i = 0
for document in corpus:
  if len(document) > 0:
    print(document)
    i = i+1
  if i == 3:
    break
```

```python
word_extractor = WordExtractor()
word_extractor.train(corpus)
word_score_table = word_extractor.extract()
```

```python
print(word_score_table["반포한"].cohesion_forward)
```

```python
print(word_score_table["반포한강"].cohesion_forward)
```

```python
print(word_score_table["반포한강공"].cohesion_forward)
```

```python
print(word_score_table["반포한강공원"].cohesion_forward)
```

```python
print(word_score_table["반포한강공원에"].cohesion_forward)
```

```python
print(word_score_table["디스"].right_branching_entropy)
```

```python
print(word_score_table["디스플"].right_branching_entropy)
```

```python
print(word_score_table["디스플레"].right_branching_entropy)
```

```python
print(word_score_table["디스플레이"].right_branching_entropy)
```

```python
from soynlp.tokenizer import LTokenizer

scores = {word:score.cohesion_forward for word, score in word_score_table.items()}
l_tokenizer = LTokenizer(scores=scores)
l_tokenizer.tokenize("국제사회와 우리의 노력들로 범죄를 척결하자", flatten=False)
```

```python
from soynlp.tokenizer import MaxScoreTokenizer

maxscore_tokenizer = MaxScoreTokenizer(scores=scores)
maxscore_tokenizer.tokenize("국제사회와우리의노력들로범죄를척결하자")
```

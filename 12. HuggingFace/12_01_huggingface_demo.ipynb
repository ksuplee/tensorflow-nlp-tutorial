```python
from transformers import pipeline

# 1) 감정 분석(모델/리비전 명시 권장)
sent = pipeline(
    task="sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    revision="714eb0f"  # 메시지에 나온 리비전 예시
)
print(sent("Hugging Face is amazing!"))
print("-" * 30)

# 2) 한국어 → 영어 번역 (Helsinki-NLP/opus-mt-ko-en)
#    태스크는 'translation_ko_to_en'로 지정하면 tokenizer 매핑이 안정적입니다.
translator = pipeline(
    task="translation_ko_to_en",
    model="Helsinki-NLP/opus-mt-ko-en",
    # device_map="auto",  # GPU 있으면 자동 할당 (원하면 주석 해제)
)
print(translator("허깅페이스는 누구나 AI를 쉽게 쓰게 해줍니다."))
```
